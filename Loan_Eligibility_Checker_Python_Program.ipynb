```python
# Input Collection

Age=int(input("Enter your age:"))
M_Income=int(input("Enter your Monthly Income:"))
Cr_Score=int(input("Enter you Credit Score:"))
Emp_Status=input("Enter you Employement Status (Employed/Unemployed):")

# Eligibility Conditions & Decision Logic 
if Age<=21:
    print("Loan Rejected: Age criteria not met")
elif M_Income<20000:
    print("Loan Rejected: Insufficient income")
elif Cr_Score<650:
    print("Loan Rejected: Low credit score")
elif Emp_Status=="Unemployed":
    print("Loan Rejected: Employment required")
else:
    print("Loan Approved")

# Loan Type Classification

if Cr_Score>=750:
    print("Premium Loan")
elif Cr_Score>=650 and Cr_Score<=749:
    print("Standard Loan")
else:
    print("")        
```
